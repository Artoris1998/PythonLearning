```python
person = {
    "name": "小華",
    "age": 20,
    "grade": 85
    }
print(person["name"])
print(person["age"])
print(person["grade"])
```

```python
person = {
    "name": "小華",
    "age": 20,
    "grade": 85
    }
person["age"] = 21

print(person["age"])
```

```python
person = {
    "name": "小華",
    "age": 20,
    "grade": 85
    }
person["city"] = "台北"

print(person)
```

```python
person = {
    "name": "小華",
    "age": 20,
    "grade": 85
    }
del person["grade"]

print(person)
```

```python
person = {
    "name": "小華",
    "age": 20,
    "grade": 85
    }
print("age" in person)
print("city" in person)
```

```python
person = {
    "name": "小華",
    "age": 20,
    "grade": 85
    }
for key in person:
  print(key)
```

```python
person = {
    "name": "小華",
    "age": 20,
    "grade": 85
    }
for key, value in person.items():
  print(key, value)
```

```python
person = {
    "name": "小明",
    "age": 18,
    "grade": 90
    }
person["age"] = 19
person["city"] = "台北"
del person["grade"]
for key, value in person.items():
  print(key, value)
print(person["name"])
```

```python
drinks = {
    "咖啡": 10,
    "紅茶": 15,
    "奶茶": 8
    }
drinks["咖啡"] = 12
drinks["綠茶"] = 20
del drinks["紅茶"]
for key, value in drinks.items():
  print(key, value)
print("奶茶" in drinks)
```

```python
employee = {
    "name": "小華",
    "age": 25,
    "salary": 32000
    }
employee["age"] = 26
employee["salary"] = 35000
employee["department"] = "門市部"

for key, value in employee.items():
  print(key, value)

print(employee["name"])
print("department" in employee)
```
