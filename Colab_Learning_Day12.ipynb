```python
person = {
    "name": "小華",
    "age": 20,
    "city": "台北"
    }

print(person.get("name"))
print(person.get("grade", 0))
print(person.keys())
print(person.values())
person.update({
    "age": 21,
    "job": "工程師"
    })
for key, value in person.items():
  print(key, value)
```

```python
drinks = {
    "咖啡": 10,
    "紅茶": 15,
    "奶茶": 8
    }

print(drinks.get("咖啡"))
print(drinks.get("綠茶", 0))
drinks.update({
    "紅茶": 20,
    "綠茶": 12
    })
result = drinks.pop("奶茶")

for key, value in drinks.items():
  print(key, value)
```

```python
students = {
    "小明": 85,
    "小華": 90,
    "小美": 78
    }
print(students.get("小華"))
students.update({
    "小美": 88,
    "小強": 92
    })
result = students.pop("小明")
print(result)
print(students.keys())
print(students.values())
for key, value in students.items():
  print(key, value)
```
