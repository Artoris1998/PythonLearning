```python
students = {
    "student1": {
        "name": "小明",
        "scores": {
            "Python": 85,
            "英文": 78,
            "數學": 90
        }
    },
    "student2": {
        "name": "小華",
        "scores": {
            "Python": 92,
            "英文": 88,
            "數學": 75
        }
    }
}

students["student1"]["scores"]["Python"] = 95
students["student1"]["scores"]["物理"] = 88
students["student2"]["scores"]["化學"] = 82

for key, value in students.items():
  print("學生姓名：", value["name"])
  print("Python成績：", value["scores"]["Python"])
  print("英文成績：", value["scores"]["英文"])
  print("數學成績：", value["scores"]["數學"])
  print("----------")
```
