```python
def my_name():
  print("我是Python學習者")
my_name()
```

```python
def my_favorite(food):
  print("我喜歡吃", food)
my_favorite("牛肉麵")
```

```python
def my_info(name, job):
  print("我的名字是", name)
  print("我的工作是", job)
my_info("小華", "超商店員")
```

```python
def multiply(a, b):
  return a * b

result = multiply(5, 6)

print(result)
```

```python
def my_drink(drink = "紅茶"):
  print("我今天喝",drink)
my_drink()
my_drink("咖啡")
```

```python
def math_operation(a, b):
  return a * b, a / b
result1, result2 = math_operation(12, 3)

print(result1)
print(result2)
```

```python
def calculate_price(price, quantity):
  return price * quantity

total = calculate_price(35, 3)

print("總金額:", total)
```

```python
def check_score(score):
  if score >= 60:
    return "及格"
  else:
    return "不及格"
result = check_score(75)

print(result)
```

```python
def discount_price(price):
  if price >= 1000:
    return price * 0.8
  else:
    return price
result = discount_price(1500)
print(result)
```

```python
def calculate_average(a, b, c):
  return (a + b +c) / 3
average = calculate_average(80, 90, 70)

print(average)
```

```python
def calculate(a, b):
  return a + b, a * b

total, product = calculate(10, 5)

print("加法結果:", total)
print("乘法結果:", product)
```
